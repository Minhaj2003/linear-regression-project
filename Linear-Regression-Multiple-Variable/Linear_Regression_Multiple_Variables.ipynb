```python
import pandas as pd
import numpy as np
import math
from sklearn import linear_model
```

```python
df = pd.read_csv(r"C:\Users\minha\OneDrive\Desktop\Linear_Regression_Multiple_Variable\hiring.csv")
df
```

```python
# STEP 1: Convert 'experience' from words → numbers

experience_map = {
    "zero": 0,
    "one": 1,
    "two": 2,
    "three": 3,
    "five": 5,
    "seven": 7,
    "ten": 10,
    "eleven": 11
}

# Convert text to numeric (unknown values become NaN)
df['experience'] = df['experience'].map(experience_map)


# STEP 2: Fill missing values in 'experience'

# Calculate median (ignores NaN automatically)
median_experience = df['experience'].median()

# Convert to integer (safe)
median_experience = int(median_experience)

# Fill missing values
df['experience'] = df['experience'].fillna(median_experience)


# STEP 3: Fill missing values in 'test_score(out of 10)'

median_test_score = df['test_score(out of 10)'].median()

df['test_score(out of 10)'] = df['test_score(out of 10)'].fillna(median_test_score)


# STEP 4: Train the model

from sklearn.linear_model import LinearRegression

reg = LinearRegression()

# Fit with multiple features
reg.fit(
    df[['experience', 'test_score(out of 10)', 'interview_score(out of 10)']],
    df['salary($)']
)


# STEP 5: Prediction

# Example input: 2 years experience, 9 test score, 6 interview score
prediction = reg.predict([[2, 9, 6]])

print("Predicted Salary:", prediction[0])


# STEP 6: Check model parameters

print("Coefficients:", reg.coef_)
print("Intercept:", reg.intercept_)
```
